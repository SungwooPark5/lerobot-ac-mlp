#### act base
0 success/ no correction 
1 fail/ no correction, boundary jump
2,4,6,9 fail/ correction 있는데 실패+boundary jump -> correction도 하고 계속 새로 시도함. 그러나 방향을 잘 못잡음
3,7,8 success/ correction 방향이 살짝 틀어짐

전체적으로 떨림은 없음, 튀는 건 있음. 거의 방향을 잘 못 잡아서 실패함

```python
from IPython.display import HTML, display
import base64
from pathlib import Path

video_dir = Path("/home1/eunji24/lerobot_project/lerobot-ac-mlp/outputs/eval/act_ens_base/videos/aloha_0")

ep = 3
video_path = video_dir / f"eval_episode_{ep}.mp4"

print(f"\n--- episode {ep} ---")

if not video_path.exists():
    print(f"파일 없음: {video_path}")
else:
    with open(video_path, "rb") as f:
        video_b64 = base64.b64encode(f.read()).decode()

    display(HTML(f"""
    <div style="margin-bottom:24px;">
        <p><b>eval_episode_{ep}.mp4</b></p>
        <video width="800" controls>
            <source src="data:video/mp4;base64,{video_b64}" type="video/mp4">
        </video>
    </div>
    """))
```
