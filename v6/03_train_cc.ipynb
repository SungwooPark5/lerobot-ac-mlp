```python
import sys, os
from pathlib import Path

GPU_OFFSET = 0  # ← only line to change (A4cc → GPU_OFFSET+0, P1cc → GPU_OFFSET+1)

sys.path.insert(0, str(Path(".").resolve()))
import common_v6 as v6

os.environ.setdefault("MUJOCO_GL", "osmesa")
os.environ.setdefault("PYOPENGL_PLATFORM", "osmesa")

TAGS = ["A4cc", "P1cc"]

# Verify model configs
for tag in TAGS:
    policy_type, lr, chunk_size, extra_flags, use_cp = v6.MODEL_CONFIGS[tag]
    print(f"{tag}: policy={policy_type}, chunk_size={chunk_size}, use_chunk_pairs={use_cp}, extra={extra_flags}")
```

```python
v6.print_training_status(TAGS)
```
