- 이 seed는 **eval 없음**(EVAL_FREQ=0 → 무거운 eval env 생성도 스킵, 순수 오프라인 학습, 체크포인트만). 필요하면 나중에 `02_eval`.
- 두 머신 쓰면 다른 seed 노트북을 다른 머신에서 동시에 (예: 머신A=seed0, 머신B=seed1).

```python
import sys, os
from pathlib import Path
sys.path.insert(0, str(Path.cwd()))
import common_final as cf
print('FINAL_TAGS:', cf.FINAL_TAGS, '| PRIMARY_SIM:', cf.PRIMARY_SIM)
```

```python
SEED = 1
TASK = cf.PRIMARY_SIM      # 'libero_10'
# seed != 0: eval 끔 → eval_freq=0 (eval env 도 안 만듦)
cf.v23.EVAL_FREQ = 0
print('eval OFF (EVAL_FREQ=0) — 체크포인트만 저장')
gpus = cf.available_gpus(list(range(4)))
print('seed 1 | GPU:', gpus, '| TASK:', TASK)
```
